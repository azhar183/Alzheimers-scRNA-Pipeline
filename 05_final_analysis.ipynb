```python
import scanpy as sc

# Setup for Jupyter
sc.settings.verbosity = 3
sc.set_figure_params(dpi=100)

# Load the heavy processed data (Takes 1 second)
adata = sc.read_h5ad('integrated_APOE3_APOE4.h5ad')
print("Data Loaded Successfully!")
```

```python
# Graph screen par hi dikhega aur wspace se overlap theek ho jayega
sc.pl.umap(adata, color=['Condition', 'leiden'], wspace=0.5)
```

```python
# 1. Pehle math calculate karein (DGE)
sc.tl.rank_genes_groups(adata, groupby='Condition', method='wilcoxon')

# 2. Phir uska Dot Plot banayein!
sc.pl.rank_genes_groups_dotplot(adata, n_genes=5)
```
